```python
import matplotlib.pyplot as plt

metrics = ["F1", "Edit Success", "WER", "ROUGE", "End-to-End"]
values = [1.0, 1.0, 0.158, 0.673, 1.0]

plt.figure()
plt.bar(metrics, values)

plt.xlabel("Evaluation Metrics")
plt.ylabel("Score")
plt.title("Overall System Performance")

plt.savefig("figure1_performance.png", dpi=300)
plt.show()
```

```python
videos = ["Short Video", "Performance", "Educational"]

f1_scores = [1.0, 1.0, 1.0]
rouge_scores = [0.65, 0.67, 0.70]  # approx per case

x = range(len(videos))

plt.figure()

plt.plot(x, f1_scores, marker='o', label="F1 Score")
plt.plot(x, rouge_scores, marker='o', label="ROUGE Score")

plt.xticks(x, videos)
plt.xlabel("Video Type")
plt.ylabel("Score")
plt.title("Performance Across Different Video Types")

plt.legend()

plt.savefig("figure3_comparison.png", dpi=300)
plt.show()
```

```python
videos = ["Short Video", "Performance Video"]
wer_scores = [0.158, 0.158]

plt.figure()
plt.bar(videos, wer_scores)

plt.xlabel("Video Type")
plt.ylabel("WER")
plt.title("Word Error Rate Across Video Types")

plt.savefig("figure4_wer.png", dpi=300)
plt.show()
```
